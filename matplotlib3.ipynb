```python
import matplotlib.pyplot as plt
import numpy as np

x=np.linspace(0,10,100)
y=np.sin(x)

plt.figure(figsize=(8,6))
plt.plot(x,y)


plt.title('Sinüs fonksiyon grafiği',fontsize=18)
plt.xlabel('X değeri',fontsize=14)
plt.ylabel('sin(x)',fontsize=14)
plt.grid(True)

plt.show()
```
